## 4. Display corpus statistics and five cosine similarities
Cosine similarity measures contextual closeness in this corpus; it does not necessarily mean that two words are synonyms.

```python
import json
from pathlib import Path

experiment = json.loads(Path('experiment.json').read_text(encoding='utf-8'))
print('Corpus sentences:', experiment['sentence_count'])
print(f"Corpus tokens: {experiment['token_count']:,}")
print('Vocabulary size:', experiment['vocabulary_size'])
print(f"\n{'Word 1':<12} {'Word 2':<12} Cosine similarity")
for row in experiment['five_reported_pairs']:
    print(f"{row['word_1']:<12} {row['word_2']:<12} {row['cosine_similarity']:.4f}")
```
